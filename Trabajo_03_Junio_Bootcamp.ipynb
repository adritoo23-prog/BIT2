```python
df= pd.read_csv('retail_fraud_detection_100k.csv')
df.head()
```

```python
# conteo de filas y columnas
df.shape
```

```python
#datos nulos

df.isnull().sum()
```

```python
#datos duplicados

df.duplicated().sum()
```

```python
#resultado limpieza de datos

df.info()
```

```python
df.describe(include="all")
```

```python
#quitar transaction_amount = 0

df=df[df['transaction_amount'] > 0]
df.head()
```

```python
df.describe(include="all")
```

```python
# ver todos los valores de transaction_amount
df.groupby('transaction_amount').count()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
#agrupar Location por transaction_id y suman de transaction_amount

dfcopy = df.groupby('location')['transaction_amount'].sum().sort_values(ascending=False)
dfcopy
```

PREGUNTAS

```python
#Cual es el país que mayor monto de ventas tiene?

fig, ax = plt.subplots(figsize=(5,5))
ax.bar(dfcopy.index,dfcopy.values,color='green',edgecolor='black')
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[0],color='black')
ax.set_xlabel('location')
ax.set_ylabel('transaction_amount')
ax.set_ylim(dfcopy.values.min() * 0.9999,dfcopy.values.max() * 1.0001)
ax.set_title('Transacciones por Pais')
plt.tight_layout
plt.show()

```
